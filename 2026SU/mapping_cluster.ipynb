```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# --- 1. Load your cluster data ---
df = pd.read_csv("data/cluster_labels.csv")
# Expected columns: county name + cluster label
# e.g.: df has 'county' and 'cluster' columns

# --- 2. Load Texas county shapefile from Census ---
url = "https://raw.githubusercontent.com/plotly/datasets/master/geojson-counties-fips.json"
texas = gpd.read_file(
    "https://www2.census.gov/geo/tiger/GENZ2021/shp/cb_2021_us_county_5m.zip"
)
texas = texas[texas['STATEFP'] == '48']  # FIPS code for Texas

# --- 3. Standardize county names for merging ---
# Strip "County" suffix and lowercase for matching
texas['county'] = texas['NAME'].str.strip()
df['county'] = df['county'].str.replace(' County', '', case=False).str.strip()

# --- 4. Merge cluster labels into shapefile ---
merged = texas.merge(df, on='county', how='left')

# --- 5. Plot ---
n_clusters = merged['cluster'].nunique()
cmap = plt.cm.get_cmap('tab10', n_clusters)

fig, ax = plt.subplots(figsize=(14, 10))
merged.plot(
    column='cluster',
    ax=ax,
    cmap='tab10',
    legend=True,
    edgecolor='white',
    linewidth=0.4,
    missing_kwds={'color': 'lightgray', 'label': 'No data'}
)

ax.set_title("Texas Counties by Cluster", fontsize=16, pad=12)
ax.set_axis_off()
plt.tight_layout()
plt.savefig("texas_clusters.png", dpi=150, bbox_inches='tight')
plt.show()
```
