### Finding Common Predictors among the models

```r
#common predictors among all models - Classification

#grabing the columns of each dataframe that contain the predictors
dtree <- dtree_vars |> pull(1)
rforest <- rforest_vars |> pull(1)
boost <- boost_vars |> pull(2)
boost_NA <- boostNA_vars |> pull(2)
rforest_reg <- rforest_reg_vars |> pull(1)

#creates a list of all the predictors from all the dataframes; used to find common predictors later
all_vars <- list(dtree, rforest, boost, boost_NA, rforest_reg)
common_vars <- Reduce(intersect, all_vars)

print(common_vars)
length(common_vars)
```

```r
rforest <- rforest_vars |> pull(1)
rforest_reg <- rforest_reg_vars |> pull(1)

all_vars <- list(rforest, rforest_reg)
common_rf <- Reduce(intersect, all_vars )


diff <- setdiff(union(rforest, rforest_reg), intersect(rforest, rforest_reg))
length(diff)

common_rf
length(common_rf)
```

```r
#common predictors between dtree and rforest
dtree <- dtree_vars |> pull(1)
rforest <- rforest_vars |> pull(1)

dec_for_common <- intersect(dtree, rforest)
dec_for_common
length(dec_for_common)
```
