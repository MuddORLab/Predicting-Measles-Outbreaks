```r
library(readr)
library(dplyr)
```

```r
#Classification model - key variables

dtree_vars <- read_csv("data/results/merged_w_svi/important_vars_4dtree.csv")

rforest_vars <- read_csv("data/results/merged_w_svi/important_vars_4rforest.csv")

boost_vars <- read_csv("data/results/merged_w_svi/important_vars_4boost.csv")
```

```r
boost_vars_NA <- read_csv("data/results/important_vars_xgboost_w_NA.csv")
```

```r
names(rforest_vars)
boost_vars$Feature
```

```r
dtree_vars_vec <- c(dtree_vars$Variable)
print("dtree predictors")
dtree_vars_vec

rforest_vars_vec <- c(rforest_vars$...1)
print("rforest predictors")
rforest_vars_vec

boost_vars_vec <- c(boost_vars$Feature)
print("boost predictors")
boost_vars_vec

boost_vars_NA_vec <- c(boost_vars_NA$Feature)
print("boost w NAs predictors")
boost_vars_NA_vec
```

```r
print("boost length")
length(boost_vars_vec)

print("boost NA length")
length(boost_vars_NA_vec)
```

```r
#common predictors among all models

#grabing the columns of each dataframe that contain the predictors
dtree <- dtree_vars |> pull(1)
rforest <- rforest_vars |> pull(1)
boost <- boost_vars |> pull(2)
boost_NA <- boost_vars_NA |> pull(2)

#creates a list of all the predictors from all the dataframes; used to find common predictors later
all_vars <- list(dtree, rforest, boost, boost_NA)
common_vars <- Reduce(intersect, all_vars)

print(common_vars)
length(common_vars)
```

```r
#common predictors between dtree and rforest
dtree <- dtree_vars |> pull(1)
rforest <- rforest_vars |> pull(1)

dec_for_common <- intersect(dtree, rforest)
dec_for_common
length(dec_for_common)
```

```r
#common predictors between boost and boost with NAs
boost <- boost_vars |> pull(2)
boost_NA <- boost_vars_NA |> pull(2)

boost_common_vars <- intersect(boost, boost_NA)
boost_common_vars
length(boost_common_vars)
```
